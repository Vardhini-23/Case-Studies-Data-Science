```python
def grade_category(grade):
    if grade <= 4:
        return "Low"
    elif grade <= 7:
        return "Medium"
    else:
        return "High"

df1["performance_category"] = df1["teacher_grade"].apply(grade_category)

print(df1["performance_category"].value_counts())

print("\nPercentages:")
print(df1["performance_category"].value_counts(normalize=True) * 100)
```

```python
# Make sure text columns contain strings
df1["student_answer"] = df1["student_answer"].fillna("").astype(str)
df1["ideal_answer"] = df1["ideal_answer"].fillna("").astype(str)

# Length of student's answer
df1["answer_word_count"] = df1["student_answer"].apply(
    lambda x: len(x.split())
)

# Length of ideal answer
df1["ideal_word_count"] = df1["ideal_answer"].apply(
    lambda x: len(x.split())
)

# Difference in length
df1["word_count_difference"] = abs(
    df1["answer_word_count"] - df1["ideal_word_count"]
)

print(df1[
    [
        "student_answer",
        "answer_word_count",
        "ideal_word_count",
        "word_count_difference",
        "performance_category"
    ]
].head())
```

```python
print(df1[
    [
        "answer_word_count",
        "ideal_word_count",
        "word_count_difference"
    ]
].describe())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

similarities = []

for student_answer, ideal_answer in zip(
    df1["student_answer"],
    df1["ideal_answer"]
):
    vectorizer = TfidfVectorizer()

    try:
        vectors = vectorizer.fit_transform(
            [student_answer, ideal_answer]
        )

        similarity = cosine_similarity(
            vectors[0],
            vectors[1]
        )[0][0]

    except ValueError:
        similarity = 0

    similarities.append(similarity)

df1["answer_similarity"] = similarities

print(df1[
    [
        "student_answer",
        "ideal_answer",
        "answer_similarity",
        "performance_category"
    ]
].head())

print("\nSimilarity statistics:")
print(df1["answer_similarity"].describe())
```

```python
print(
    df1.groupby("performance_category")[
        ["answer_word_count",
         "word_count_difference",
         "answer_similarity"]
    ].mean()
)
```

```python
features = [
    "answer_word_count",
    "ideal_word_count",
    "word_count_difference",
    "answer_similarity"
]

X = df1[features]
y = df1["performance_category"]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())

print("\nMissing values:")
print(X.isnull().sum())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])

print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)

rf_model.fit(X_train, y_train)

rf_predictions = rf_model.predict(X_test)

print("Random Forest training complete.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

print("Random Forest Results")

print("Accuracy:",
      accuracy_score(y_test, rf_predictions))

print("Precision:",
      precision_score(
          y_test,
          rf_predictions,
          average="weighted"
      ))

print("Recall:",
      recall_score(
          y_test,
          rf_predictions,
          average="weighted"
      ))

print("F1 Score:",
      f1_score(
          y_test,
          rf_predictions,
          average="weighted"
      ))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        rf_predictions
    )
)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Scaling complete.")
```

```python
from sklearn.svm import SVC

svm_model = SVC(
    kernel="rbf",
    class_weight="balanced",
    random_state=42
)

svm_model.fit(X_train_scaled, y_train)

svm_predictions = svm_model.predict(X_test_scaled)

print("SVM training complete.")
```

```python
print("SVM Results")

print("Accuracy:",
      accuracy_score(y_test, svm_predictions))

print("Precision:",
      precision_score(
          y_test,
          svm_predictions,
          average="weighted"
      ))

print("Recall:",
      recall_score(
          y_test,
          svm_predictions,
          average="weighted"
      ))

print("F1 Score:",
      f1_score(
          y_test,
          svm_predictions,
          average="weighted"
      ))

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        svm_predictions
    )
)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

ConfusionMatrixDisplay.from_predictions(
    y_test,
    rf_predictions
)

plt.title("Random Forest - Dataset 1")
plt.tight_layout()
plt.show()
```

```python
df2 = results.merge(
    students[
        [
            "STUDENT_ID",
            "GENDER",
            "HARMOS_GRADE",
            "SCHOOL_TYPE",
            "AGE"
        ]
    ],
    on="STUDENT_ID",
    how="left"
)

print("Original RESULTS shape:", results.shape)
print("Merged Dataset 2 shape:", df2.shape)

print("\nSelected student variables missing values:")
print(
    df2[
        [
            "GENDER",
            "HARMOS_GRADE",
            "SCHOOL_TYPE",
            "AGE"
        ]
    ].isnull().sum()
)

print("\nTarget distribution:")
print(df2["CORRECT"].value_counts(dropna=False))
```

```python
df2 = df2.dropna(subset=["CORRECT"]).copy()

print("Shape after removing missing CORRECT values:")
print(df2.shape)

print("\nTarget counts:")
print(df2["CORRECT"].value_counts())

print("\nTarget percentages:")
print(df2["CORRECT"].value_counts(normalize=True) * 100)
```

```python
features2 = [
    "TIME",
    "TIME_DIFF",
    "ARTEFACT_DIMENSION",
    "ARTEFACT_TYPE",
    "ARTEFACT_CATEGORY",
    "DOMAIN",
    "GENDER",
    "HARMOS_GRADE",
    "SCHOOL_TYPE",
    "AGE"
]

X2 = df2[features2].copy()
y2 = df2["CORRECT"].astype(bool)

print("Feature data shape:", X2.shape)
print("Target shape:", y2.shape)

print("\nData types:")
print(X2.dtypes)

print("\nMissing values:")
print(X2.isnull().sum())
```

```python
print("Missing values:")
print(X2.isnull().sum())
```

```python
X2 = X2.drop(columns=["DOMAIN"])

print("Updated features:")
print(X2.columns.tolist())

print("\nMissing values:")
print(X2.isnull().sum())

print("\nShape:")
print(X2.shape)
```

```python
X2_encoded = pd.get_dummies(
    X2,
    drop_first=True
)

print("Original shape:", X2.shape)
print("Encoded shape:", X2_encoded.shape)

print("\nData types:")
print(X2_encoded.dtypes.value_counts())

print("\nAny missing values:")
print(X2_encoded.isnull().sum().sum())
```

```python
from sklearn.model_selection import train_test_split

X2_train, X2_test, y2_train, y2_test = train_test_split(
    X2_encoded,
    y2,
    test_size=0.20,
    random_state=42,
    stratify=y2
)

print("Training samples:", X2_train.shape[0])
print("Testing samples:", X2_test.shape[0])

print("\nTraining target distribution:")
print(y2_train.value_counts())

print("\nTesting target distribution:")
print(y2_test.value_counts())

print("\nTesting target percentages:")
print(y2_test.value_counts(normalize=True) * 100)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf2_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)

rf2_model.fit(X2_train, y2_train)

rf2_predictions = rf2_model.predict(X2_test)

print("Dataset 2 Random Forest training complete.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

print("Dataset 2 - Random Forest Results")

print("Accuracy:",
      accuracy_score(y2_test, rf2_predictions))

print("Precision:",
      precision_score(
          y2_test,
          rf2_predictions,
          average="weighted"
      ))

print("Recall:",
      recall_score(
          y2_test,
          rf2_predictions,
          average="weighted"
      ))

print("F1 Score:",
      f1_score(
          y2_test,
          rf2_predictions,
          average="weighted"
      ))

print("\nClassification Report:")
print(
    classification_report(
        y2_test,
        rf2_predictions
    )
)
```

```python
def grade_category(grade):
    if grade <= 4:
        return "Low"
    elif grade <= 7:
        return "Medium"
    else:
        return "High"

df1["performance_category"] = df1["teacher_grade"].apply(grade_category)

print(df1["performance_category"].value_counts())

print("\nPercentages:")
print(df1["performance_category"].value_counts(normalize=True) * 100)
```

```python
# Make sure text columns contain strings
df1["student_answer"] = df1["student_answer"].fillna("").astype(str)
df1["ideal_answer"] = df1["ideal_answer"].fillna("").astype(str)

# Length of student's answer
df1["answer_word_count"] = df1["student_answer"].apply(
    lambda x: len(x.split())
)

# Length of ideal answer
df1["ideal_word_count"] = df1["ideal_answer"].apply(
    lambda x: len(x.split())
)

# Difference in length
df1["word_count_difference"] = abs(
    df1["answer_word_count"] - df1["ideal_word_count"]
)

print(df1[
    [
        "student_answer",
        "answer_word_count",
        "ideal_word_count",
        "word_count_difference",
        "performance_category"
    ]
].head())
```

```python
print(df1[
    [
        "answer_word_count",
        "ideal_word_count",
        "word_count_difference"
    ]
].describe())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

similarities = []

for student_answer, ideal_answer in zip(
    df1["student_answer"],
    df1["ideal_answer"]
):
    vectorizer = TfidfVectorizer()

    try:
        vectors = vectorizer.fit_transform(
            [student_answer, ideal_answer]
        )

        similarity = cosine_similarity(
            vectors[0],
            vectors[1]
        )[0][0]

    except ValueError:
        similarity = 0

    similarities.append(similarity)

df1["answer_similarity"] = similarities

print(df1[
    [
        "student_answer",
        "ideal_answer",
        "answer_similarity",
        "performance_category"
    ]
].head())

print("\nSimilarity statistics:")
print(df1["answer_similarity"].describe())
```

```python
features = [
    "answer_word_count",
    "ideal_word_count",
    "word_count_difference",
    "answer_similarity"
]

X = df1[features]
y = df1["performance_category"]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())

print("\nMissing values:")
print(X.isnull().sum())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])

print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)

rf_model.fit(X_train, y_train)

rf_predictions = rf_model.predict(X_test)

print("Random Forest training complete.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

print("Random Forest Results")

print("Accuracy:",
      accuracy_score(y_test, rf_predictions))

print("Precision:",
      precision_score(
          y_test,
          rf_predictions,
          average="weighted"
      ))

print("Recall:",
      recall_score(
          y_test,
          rf_predictions,
          average="weighted"
      ))

print("F1 Score:",
      f1_score(
          y_test,
          rf_predictions,
          average="weighted"
      ))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        rf_predictions
    )
)
```

### 2.1 Cross-Validation and Sampling Strategy

To assess whether the original model evaluation is stable across different
samples, Stratified 5-Fold Cross-Validation is applied. Stratification is
used because the CORRECT target is imbalanced and preserves approximately
the same class distribution in each fold.

```python
# Stratified 5-Fold Cross-Validation for Dataset 2

from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.ensemble import RandomForestClassifier

# Use the same Random Forest configuration as Task 1
rf_cv = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)

# Preserve the target class distribution in each fold
skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Evaluate using multiple metrics
cv_results = cross_validate(
    rf_cv,
    X2_encoded,
    y2,
    cv=skf,
    scoring=["accuracy", "precision_weighted",
             "recall_weighted", "f1_weighted"]
)

# Display results for each fold
cv_results_df = pd.DataFrame({
    "Fold": range(1, 6),
    "Accuracy": cv_results["test_accuracy"],
    "Precision": cv_results["test_precision_weighted"],
    "Recall": cv_results["test_recall_weighted"],
    "F1 Score": cv_results["test_f1_weighted"]
})

print(cv_results_df.round(4))

print("\nMean Cross-Validation Performance:")
print(cv_results_df.iloc[:, 1:].mean().round(4))

print("\nStandard Deviation:")
print(cv_results_df.iloc[:, 1:].std().round(4))
```

## 2.2 Model Performance Under Varying Training Set Sizes

A learning curve is used to examine how Random Forest performance changes
as the amount of training data increases.

```python
# Learning Curve - Dataset 2 Random Forest

from sklearn.model_selection import learning_curve
import numpy as np
import matplotlib.pyplot as plt

# Training set sizes to evaluate
train_sizes = np.linspace(0.1, 1.0, 10)

# Generate learning curve using stratified 5-fold CV
sizes, train_scores, validation_scores = learning_curve(
    rf_cv,
    X2_encoded,
    y2,
    train_sizes=train_sizes,
    cv=skf,
    scoring="f1_weighted",
    shuffle=True,
    random_state=42,
    n_jobs=-1
)

# Calculate mean and standard deviation
train_mean = train_scores.mean(axis=1)
train_std = train_scores.std(axis=1)

validation_mean = validation_scores.mean(axis=1)
validation_std = validation_scores.std(axis=1)

# Display numerical results
learning_results = pd.DataFrame({
    "Training Size": sizes,
    "Training F1": train_mean,
    "Validation F1": validation_mean,
    "Validation SD": validation_std
})

print(learning_results.round(4))

# Plot the learning curve
plt.figure(figsize=(8, 5))

plt.plot(
    sizes,
    train_mean,
    marker="o",
    label="Training F1"
)

plt.plot(
    sizes,
    validation_mean,
    marker="o",
    label="Validation F1"
)

plt.fill_between(
    sizes,
    validation_mean - validation_std,
    validation_mean + validation_std,
    alpha=0.15
)

plt.xlabel("Training Set Size")
plt.ylabel("Weighted F1 Score")
plt.title("Learning Curve - Random Forest (Dataset 2)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig("learning_curve_dataset2.png", dpi=300, bbox_inches="tight")
plt.show()
```

Microsoft Fairlearn is used to investigate whether model performance

```python
# Align GENDER with the existing Dataset 2 test set

gender_test = df2.loc[X2_test.index, "GENDER"]

print("Test set size:", len(X2_test))
print("Sensitive feature size:", len(gender_test))

print("\nGENDER distribution in test set:")
print(gender_test.value_counts())

print("\nMissing GENDER values:")
print(gender_test.isna().sum())
```

```python
# Fairlearn group fairness analysis - Random Forest by GENDER

from fairlearn.metrics import (
    MetricFrame,
    selection_rate,
    false_positive_rate
)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

from sklearn.ensemble import RandomForestClassifier

# Recreate and train the same Random Forest model
rf2 = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)

rf2.fit(X2_train, y2_train)

# Generate predictions on the existing test set
y2_pred_rf = rf2.predict(X2_test)

# Define fairness metrics
fairness_metrics = {
    "Accuracy": accuracy_score,
    "Precision": precision_score,
    "Recall / TPR": recall_score,
    "F1 Score": f1_score,
    "Selection Rate": selection_rate,
    "False Positive Rate": false_positive_rate
}

# Calculate metrics overall and by gender
metric_frame = MetricFrame(
    metrics=fairness_metrics,
    y_true=y2_test,
    y_pred=y2_pred_rf,
    sensitive_features=gender_test
)

print("Overall model performance:")
print(metric_frame.overall.round(4))

print("\nPerformance by GENDER:")
print(metric_frame.by_group.round(4))

print("\nAbsolute difference between groups:")
print(metric_frame.difference().round(4))

print("\nRatio between groups:")
print(metric_frame.ratio().round(4))
```

```python
# Standard Fairlearn fairness metrics

from fairlearn.metrics import (
    demographic_parity_difference,
    demographic_parity_ratio,
    equalized_odds_difference,
    equalized_odds_ratio
)

dp_difference = demographic_parity_difference(
    y2_test,
    y2_pred_rf,
    sensitive_features=gender_test
)

dp_ratio = demographic_parity_ratio(
    y2_test,
    y2_pred_rf,
    sensitive_features=gender_test
)

eo_difference = equalized_odds_difference(
    y2_test,
    y2_pred_rf,
    sensitive_features=gender_test
)

eo_ratio = equalized_odds_ratio(
    y2_test,
    y2_pred_rf,
    sensitive_features=gender_test
)

print("Demographic Parity Difference:", round(dp_difference, 4))
print("Demographic Parity Ratio:", round(dp_ratio, 4))
print("Equalized Odds Difference:", round(eo_difference, 4))
print("Equalized Odds Ratio:", round(eo_ratio, 4))
```

## 2.4 Security, Privacy and Ethical Risks

Security:   If this methodology were deployed on a larger scale, it could create security, privacy, and ethical risks. From a security perspective, the system would process educational and student-related information, so unauthorised access, data breaches, or manipulation of records could affect both students and model predictions. Secure storage, access controls, and monitoring would therefore be important. 

Privacy:    Dataset 2 contains demographic and educational attributes such as gender, age, school type, and performance-related information. Combining these attributes could increase the possibility of profiling or identifying students, so data minimisation, de-identification, and restricted access should be considered. 

Ethics: The fairness analysis indicates that similar overall prediction rates do not necessarily mean similar treatment across groups. Although demographic parity was relatively close, the Random Forest produced a higher false-positive rate for females (0.8571) than males (0.6452), with an equalized-odds difference of 0.2122. If predictions were used to make decisions affecting students, unequal error rates could disadvantage particular groups. 

Therefore, the model should be regularly audited across demographic groups and used with human oversight rather than as the sole basis for consequential decisions.
